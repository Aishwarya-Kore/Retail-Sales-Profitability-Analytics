```python
monthly_sales['Previous_Month_Sales'] = (
    monthly_sales['Total_Sales'].shift(1)
)

monthly_sales['MoM_Growth_%'] = (
    (monthly_sales['Total_Sales'] -
     monthly_sales['Previous_Month_Sales']) /
    monthly_sales['Previous_Month_Sales']
) * 100
```

```python
def discount_category(discount):
    if discount == 0:
        return 'No Discount'
    elif discount <= 0.20:
        return 'Low Discount'
    elif discount <= 0.40:
        return 'Medium Discount'
    else:
        return 'High Discount'

df_clean['Discount_Band'] = df_clean['Discount'].apply(
    discount_category
)
```

```python
discount_analysis = df_clean.groupby('Discount_Band').agg(
    Total_Sales=('Sales', 'sum'),
    Total_Profit=('Profit', 'sum'),
    Total_Quantity=('Quantity', 'sum'),
    Total_Orders=('Order ID', 'nunique'),
    Average_Discount=('Discount', 'mean')
).reset_index()
```

```python
df_clean['Shipping_Days'] = (
    df_clean['Ship Date'] - df_clean['Order Date']
).dt.days
```

```python
df_clean[['Order Date', 'Ship Date', 'Shipping_Days']].head(10)
```

```python
(df_clean['Shipping_Days'] < 0).sum()
```

```python
df_clean['Shipping_Days'].describe()
```

```python
shipping_analysis = df_clean.groupby('Ship Mode').agg(
    Total_Orders=('Order ID', 'nunique'),
    Total_Sales=('Sales', 'sum'),
    Total_Profit=('Profit', 'sum'),
    Average_Shipping_Days=('Shipping_Days', 'mean')
).reset_index()
```

```python
shipping_analysis.sort_values(
    'Average_Shipping_Days'
)
```

```python
def shipping_category(days):
    if days <= 2:
        return 'Fast'
    elif days <= 4:
        return 'Standard'
    else:
        return 'Slow'

df_clean['Shipping_Category'] = (
    df_clean['Shipping_Days'].apply(shipping_category)
)
```

```python
shipping_distribution = df_clean[
    ['Order ID', 'Shipping_Category']
].drop_duplicates().groupby(
    'Shipping_Category'
).agg(
    Total_Orders=('Order ID', 'count')
).reset_index()

shipping_distribution
```

```python
shipping_analysis.plot(
    x='Ship Mode',
    y='Average_Shipping_Days',
    kind='bar',
    figsize=(9, 5),
    legend=False
)

plt.title('Average Shipping Days by Shipping Mode')
plt.xlabel('Shipping Mode')
plt.ylabel('Average Days')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
correlation = df_clean[
    ['Sales', 'Profit', 'Quantity', 'Discount', 'Shipping_Days']
].corr()

correlation
```

```python
plt.figure(figsize=(9, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Retail Business Correlation Analysis')
plt.tight_layout()
plt.show()
```
